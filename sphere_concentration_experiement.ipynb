```python
import numpy as np
import matplotlib.pyplot as plt
from tqdm.notebook import tqdm

# -----------------------------
# Parameters
# -----------------------------
d = 10
epsilon = 1.0
n_values = np.arange(20, 161, 2)

num_trials = 5000
rng = np.random.default_rng(42)

# -----------------------------
# Theoretical bound
# -----------------------------
def theoretical_bound(n, d, epsilon):
    denominator = 2*d - 2 + (2 * epsilon * (d - 1)) / 3
    return 2*d * np.exp(- n * epsilon**2 / denominator)

# -----------------------------
# Sampling from sphere of radius sqrt(d/n)
# -----------------------------
def sample_scaled_sphere_vectors(n, d, rng):
    """
    Returns an n x d matrix X whose rows are independent samples
    from the sphere S^{d-1}_{sqrt(d/n)}.
    """
    X = rng.normal(size=(n, d))
    X /= np.linalg.norm(X, axis=1, keepdims=True)
    X *= np.sqrt(d / n)
    return X

# -----------------------------
# One Monte Carlo estimate
# -----------------------------
def empirical_probability(n, d, epsilon, num_trials, rng):
    count = 0

    for _ in range(num_trials):
        X = sample_scaled_sphere_vectors(n, d, rng)

        # Frame operator S = sum_i x_i x_i^*
        S = X.T @ X

        # Operator norm of S - I_d
        op_norm = np.linalg.norm(S - np.eye(d), ord=2)

        if op_norm >= epsilon:
            count += 1

    return count / num_trials

# -----------------------------
# Run experiment
# -----------------------------
empirical_probs = []
theoretical_probs = []

for n in tqdm(n_values, desc="Running experiment"):
    emp = empirical_probability(n, d, epsilon, num_trials, rng)
    th = theoretical_bound(n, d, epsilon)

    empirical_probs.append(emp)
    theoretical_probs.append(th)

empirical_probs = np.array(empirical_probs)
theoretical_probs = np.array(theoretical_probs)

# Since probabilities cannot exceed 1, we clip only for plotting.
theoretical_probs_clipped = np.minimum(theoretical_probs, 1)

# -----------------------------
# Plot
# -----------------------------
plt.figure(figsize=(9, 5.5))

plt.plot(
    n_values,
    empirical_probs,
    marker="o",
    linestyle="-",
    color="red",
    label=r"Empirical $P(\|S-I_d\|_{\mathrm{op}} \geq \varepsilon)$"
)

plt.plot(
    n_values,
    theoretical_probs_clipped,
    linestyle="--",
    linewidth=2,
    color="turquoise",
    label="Theoretical bound"
)

plt.title(rf"$P$ vs $n$ $(d={d}, \varepsilon={epsilon})$", fontsize=14)
plt.xlabel(r"$n$", fontsize=12)
plt.ylabel(r"$P$", fontsize=12)

plt.ylim(0, 1.05)
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```
